```python
#print all even numbers from 1 to 20

for i in range(1,21):
    if i%2==0:
        print(i)
```

```python
#print all odd number from 1 to 10

for i in range(1,21):
    if(i%2!=0):
        print(i)
```

```python
#print multiples of 5 from 5 to 50

for i in range(5,51,5):
    print(i)
```

```python
#find the sum of numbers 1 to 10

sum=0
for i in range(1,11):
    sum+=i #sum=sum+i sum=0+1
print(sum)
```

```python
#find the sum of even numbers from 1 to 100

sum=0
for i in range(1,101,1):
    if i%2==0:
        sum+=i
print(sum)
```

```python
#sum of odd numbers from 1 to 100

sum=0
for i in range(1,101):
    if i%2!=0:
        sum=sum+i
print(sum)
```

```python
#print the multiplication table of 7

for i in range(1,11):
    print(7*i)
```

```python
#take number from user and print its multiplication table

num=int(input("Enter a number:"))
for i in range(1,11):
    print(num,"x",i,"=",num*i)
```

```python
#print numbers from 1 to 100 that are divisible by both 3 and 5.
for i in range(1,101):
    if i%3==0 and i%5==0:
        print(i)
```

```python
#count even and odd numbers between 1 to 50

even=0
odd=0

for i in range(1,51):
    if i%2==0:
        even=even+1
    else:
        odd=odd+1
print("even numbers=",even)
print("odd numbers=",odd)
```

```python
#print the elements of a list in reverse order
#data=[22,33,12,32,45,13]
data=[22,33,12,32,45,13]
print(data[::-1])
```

```python
#find the largest number in a list using for loop
#data=[22,55,12,89,34,67]

data=[22,55,12,89,34,67]
largest=data[0]
for i in data:
    if i>largest:
        largest=i
print("largest number:", largest)
```

```python
#wap to count a vowel in a particular string
data="Programming"
count=0
for i in data:
    if i in "aeiou":
        count+=1
print(count)
```

```python
#wap to count no. of spaces in a particular sentence
sentence="Python is very easy"
count=0
for i in sentence:
    if i==" ":
       count+=1
print("Number of spaces:",count)
```
