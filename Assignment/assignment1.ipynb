```python
# Exercise 1: ATM Withdrawal
# Question: Write a Python program for ATM withdrawal.
# Conditions:
# - Check whether the PIN is correct.
# - If the PIN is correct, check whether the withdrawal amount is a multiple of 100.
# - If it is, check whether the balance is sufficient.
# - If sufficient, withdraw the amount.
# - Otherwise, print "Insufficient Balance".
# - If the PIN is incorrect, print "Invalid PIN".


pin = int(input("Enter your ATM PIN: "))

balance = 10000
correct_pin = 1234

if pin == correct_pin:
    amount = int(input("Enter withdrawal amount: "))

    if amount % 100 == 0:
        if amount > 0:
            if amount <= balance:
                balance = balance - amount
                print("Withdrawal Successful")
                print("Remaining Balance:", balance)
            else:
                print("Insufficient Balance")
        else:
            print("Invalid Withdrawal Amount")
    else:
        print("Amount must be a multiple of 100")
else:
    print("Invalid PIN")
```

```python
# Exercise 2: Student Result System
# Question: Write a program that takes marks of 3 subjects.
# Conditions:
# - If every subject has marks greater than or equal to 40, calculate the percentage.
# - Percentage ≥ 75: Distinction
# - Percentage ≥ 60: First Division
# - Percentage ≥ 50: Second Division
# - Otherwise: Pass
# - If any subject has marks below 40, print Fail.


m1 = float(input("Enter marks of Subject 1: "))
m2 = float(input("Enter marks of Subject 2: "))
m3 = float(input("Enter marks of Subject 3: "))

if m1 >= 40 and m2 >= 40 and m3 >= 40:
    percentage = (m1 + m2 + m3) / 3

    print("Percentage:", percentage)

    if percentage >= 75:
        print("Distinction")
    elif percentage >= 60:
        print("First Division")
    elif percentage >= 50:
        print("Second Division")
    else:
        print("Pass")
else:
    print("Fail")
```

```python
# Exercise 3: Bank Loan Eligibility
# Question: Write a Python program to check bank loan eligibility.
# Conditions:
# - Age must be ≥ 21.
# - If age is valid, salary must be ≥ ₹25,000.
# - If salary is valid:
#   - Credit score ≥ 750: Loan approved.
#   - Credit score ≥ 650: Loan approved with higher interest.
#   - Otherwise: Loan rejected.


age = int(input("Enter your age: "))

if age >= 21:
    salary = float(input("Enter your monthly salary: "))

    if salary >= 25000:
        credit_score = int(input("Enter your credit score: "))

        if credit_score >= 750:
            print("Loan Approved")
        elif credit_score >= 650:
            print("Loan Approved with Higher Interest")
        else:
            print("Loan Rejected")
    else:
        print("Salary is too low for the loan")
else:
    print("Age must be at least 21")
```

```python
# Exercise 4: E-Commerce Discount
# Question: Write a Python program to calculate an online shopping discount.
# Conditions:
# - If cart amount ≥ ₹5,000:
#   - Premium member: 20% discount.
#   - Otherwise: 10% discount.
# - If cart amount is below ₹5,000:
#   - Premium member: 5% discount.
#   - Otherwise: No discount.
# - Print the final amount after applying the discount.

amount = float(input("Enter cart amount: "))
premium = input("Are you a premium member? (yes/no): ")

if amount >= 5000:
    if premium == "yes":
        discount = 20
    else:
        discount = 10
else:
    if premium == "yes":
        discount = 5
    else:
        discount = 0

discount_amount = amount * discount / 100
final_amount = amount - discount_amount

print("Discount:", discount, "%")
print("Discount Amount:", discount_amount)
print("Final Amount:", final_amount)
```

```python
# Exercise 5: Login + Role-Based Access
# Question: Create a login system.
# Conditions:
# - Check username first.
# - If the username is correct, check the password.
# - If the password is correct, check the role:
#   - "admin" → Full Access
#   - "teacher" → Teacher Dashboard
#   - "student" → Student Dashboard
#   - Anything else → Invalid Role
# - Wrong password → "Wrong Password"
# - Wrong username → "Invalid Username"

correct_username = "vaishnavi"
correct_password = "12345"

username = input("Enter username: ")

if username == correct_username:
    password = input("Enter password: ")

    if password == correct_password:
        role = input("Enter your role: ")

        if role == "admin":
            print("Full Access")
        elif role == "teacher":
            print("Teacher Dashboard")
        elif role == "student":
            print("Student Dashboard")
        else:
            print("Invalid Role")
    else:
        print("Wrong Password")
else:
    print("Invalid Username")
```

```python
# Exercise 6: Movie Ticket Booking
# Question: Create a movie ticket booking system.
# Conditions:
# - Check whether seats are available.
# - If seats are available, check customer age:
#   - Age below 5 → Free
#   - Age from 5 to 12 → 50% discount
#   - Age above 60 → 30% discount
#   - Otherwise → Full price
# - If no seats are available, print "House Full".

seats = int(input("Enter available seats: "))

if seats > 0:
    age = int(input("Enter customer age: "))
    ticket_price = 200

    if age < 0:
        print("Invalid Age")
    elif age < 5:
        discount = 100
    elif age <= 12:
        discount = 50
    elif age > 60:
        discount = 30
    else:
        discount = 0

    if age >= 0:
        final_price = ticket_price - (ticket_price * discount / 100)

        print("Ticket Price:", ticket_price)
        print("Discount:", discount, "%")
        print("Final Ticket Price:", final_price)

        seats = seats - 1
        print("Remaining Seats:", seats)
else:
    print("House Full")
```
