```python
#Find numbers greater than 50 and store all the numbers into a new list.

numbers = [20, 55, 30, 80, 45, 90, 10]
result = [x for x in numbers if x > 50]
print(result)
```

```python
#Find numbers between 20 and 50 and store all the numbers into a new list.

numbers = [10, 25, 35, 60, 45, 15, 50, 70]
result = [x for x in numbers if 20 <= x <= 50]
print(result)
```

```python
#Find numbers divisible by both 3 and 5 and store all the numbers into a new list.

numbers = [10, 15, 20, 30, 45, 50, 60, 75]
result = [x for x in numbers if x % 3 == 0 and x % 5 == 0]
print(result)
```

```python
#Find names having more than 4 characters and store the result into a new list.

names = ["Amit", "Rahul", "Neha", "Rohit", "Sneha"]
result = [name for name in names if len(name) > 4]
print(result)
```

```python
#Find names starting with "A" and store the result into a new list.

names = ["Amit", "Rahul", "Ankit", "Neha", "Arjun", "Rohit"]
result = [name for name in names if name.startswith("A")]
print(result)
```

```python
#Extract vowels from a string and store the result into a new list.

text = "programming"
result = [x for x in text if x in "aeiou"]
print(result)
```

```python
#Extract digits from a string and store the result into a new list.

data = "Python123Programming456"
result = [x for x in data if x.isdigit()]
print(result)
```

```python
#Replace every even number with "Even" and every odd number with "Odd" and store the result into a new list.

numbers = [10, 15, 20, 25, 30]
result = ["Even" if x % 2 == 0 else "Odd" for x in numbers]
print(result)
```

```python
#Replace negative numbers with 0 and store the result into a new list

numbers = [10, -5, 20, -8, 30, -2]
result = [0 if x < 0 else x for x in numbers]
print(result)
```

```python
#Give a 10% discount to products costing ₹1000 or more, otherwise give 5%, and store the result into a new list.

prices = [500, 1200, 800, 2000, 1500]
result = [x * 0.90 if x >= 1000 else x * 0.95 for x in prices]
print(result)
```

```python
#Employee Management: Find employees earning more than ₹50,000.

employees = [
    {"name": "Amit", "salary": 45000},
    {"name": "Rahul", "salary": 60000},
    {"name": "Neha", "salary": 75000},
    {"name": "Rohit", "salary": 35000}
]

result = [x["name"] for x in employees if x["salary"] > 50000]

print(result)
```

```python
#E-Commerce: Find products costing more than ₹1000.

prices = [499, 1299, 799, 2499, 999, 1599]
result = [x for x in prices if x > 1000]
print(result)
```

```python
#E-Commerce Discount: Give a 20% discount only to products costing more than ₹2000.

prices = [500, 1200, 2500, 3000, 1500, 5000]
result = [x * 0.80 for x in prices if x > 2000]
print(result)
```

```python
#Bank Transactions: Find all deposits.

transactions = [5000, -2000, 3000, -1000, 7000, -500]
deposits = [x for x in transactions if x > 0]
print(deposits)
```

```python
#Find names of students who scored 60 or more.

students = [
    {"name": "Amit", "marks": 75},
    {"name": "Rahul", "marks": 45},
    {"name": "Neha", "marks": 90},
    {"name": "Rohit", "marks": 55}
]

result = [x["name"] for x in students if x["marks"] >= 60]

print(result)
```

```python
#Find names of employees from the IT department.

employees = [
    {"name": "Amit", "department": "IT"},
    {"name": "Rahul", "department": "HR"},
    {"name": "Neha", "department": "IT"},
    {"name": "Rohit", "department": "Sales"}
]

result = [x["name"] for x in employees if x["department"] == "IT"]

print(result)
```

```python
#Find employees who earn more than ₹60,000 and belong to IT.
employees = [
    {"name": "Amit", "department": "IT", "salary": 55000},
    {"name": "Rahul", "department": "HR", "salary": 70000},
    {"name": "Neha", "department": "IT", "salary": 80000},
    {"name": "Rohit", "department": "IT", "salary": 45000}
]

result = [
    x["name"] for x in employees
    if x["salary"] > 60000 and x["department"] == "IT"
]

print(result)
```
