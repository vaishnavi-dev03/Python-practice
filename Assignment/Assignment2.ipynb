```python
# Exercise 7: Electricity Bill

# Question:
# Write a Python program to calculate an electricity bill
# based on the number of units consumed.
#
# Conditions:
# 0-100 units    -> Rs. 5/unit
# 101-200 units  -> Rs. 7/unit
# 201-300 units  -> Rs. 10/unit
# Above 300      -> Rs. 15/unit
#
# Additionally:
# If bill > Rs. 3000, add a 10% surcharge.

units = int(input("Enter electricity units: "))

if units <= 100:
    bill = units * 5
elif units <= 200:
    bill = units * 7
elif units <= 300:
    bill = units * 10
else:
    bill = units * 15

if bill > 3000:
    surcharge = bill * 0.10
    bill = bill + surcharge

print("Total Electricity Bill: Rs.", bill)
```

```python
# Exercise 8: Employee Bonus

# Question:
# Write a Python program to calculate an employee's bonus
# based on experience, salary and performance.
#
# Conditions:
# If experience >= 1 year and salary >= Rs. 50,000:
#     Performance >= 90 -> 20% bonus
#     Performance >= 75 -> 10% bonus
#     Otherwise -> 5% bonus
#
# If salary < Rs. 50,000:
#     Performance >= 90 -> 15% bonus
#     Otherwise -> 5% bonus

experience = int(input("Enter experience in years: "))
salary = float(input("Enter salary: Rs. "))
performance = int(input("Enter performance score: "))

if experience >= 1:
    if salary >= 50000:
        if performance >= 90:
            bonus_percent = 20
        elif performance >= 75:
            bonus_percent = 10
        else:
            bonus_percent = 5
    else:
        if performance >= 90:
            bonus_percent = 15
        else:
            bonus_percent = 5

    bonus = salary * bonus_percent / 100

    print("Bonus Percentage:", bonus_percent, "%")
    print("Bonus Amount: Rs.", bonus)
    print("Total Salary: Rs.", salary + bonus)

else:
    print("Minimum 1 year experience required")
```

```python
# Exercise 9: Driving License

# Question:
# Write a Python program to check whether a person
# is eligible for a driving license.
#
# Conditions:
# Age must be >= 18.
# Check whether a learner license is available.
# Driving test score must be >= 60.

age = int(input("Enter your age: "))

if age >= 18:
    learner_license = input(
        "Do you have a learner license? (yes/no): "
    ).lower()

    if learner_license == "yes":
        score = int(input("Enter driving test score: "))

        if score >= 60:
            print("Eligible for Driving License")
        else:
            print("Driving Test Failed")

    else:
        print("Learner License Required")

else:
    print("Not Eligible: Age must be 18 or above")
```

```python
# Exercise 10: ATM - Difficult Challenge

# Question:
# Write a Python program for ATM withdrawal.
#
# Conditions:
# 1. PIN must be correct.
# 2. Account must be active.
# 3. Withdrawal amount must be positive.
# 4. Amount must be a multiple of Rs. 100.
# 5. Balance must be sufficient.
# 6. After withdrawal, minimum Rs. 500 balance must remain.
# 7. Amount > Rs. 20,000 requires extra verification.

correct_pin = 1234
balance = 30000
account_active = True

pin = int(input("Enter PIN: "))

if pin == correct_pin:
    if account_active:
        amount = int(input("Enter withdrawal amount: "))

        if amount > 0:
            if amount % 100 == 0:
                if amount > 20000:
                    verification = input(
                        "Extra verification (yes/no): "
                    ).lower()
                else:
                    verification = "yes"

                if verification == "yes":
                    if amount <= balance - 500:
                        balance = balance - amount
                        print("Withdrawal Successful")
                        print("Remaining Balance: Rs.", balance)
                    else:
                        print("Minimum Rs. 500 balance must remain")
                else:
                    print("Verification Failed")
            else:
                print("Amount must be a multiple of Rs. 100")
        else:
            print("Amount must be positive")
    else:
        print("Account is inactive")
else:
    print("Invalid PIN")
```
